```python
# ── Imports ──────────────────────────────────────────────────────────────────
import sys
from pathlib import Path

import pandas as pd
import altair as alt
import vl_convert as vlc

alt.data_transformers.disable_max_rows()
alt.renderers.enable('svg')

CLEAN_FILE = Path('../../data/processed/hotel_bookings_clean.csv')
NAMES_FILE = Path('../../data/processed/country_names.csv')
OUTPUT_DIR = Path('interaction_plan_output')
OUTPUT_DIR.mkdir(exist_ok=True)

print('Python', sys.version.split()[0], '| pandas', pd.__version__, '| altair', alt.__version__,
      '| Vega-Lite', alt.SCHEMA_VERSION, '| vl-convert', vlc.__version__)
```

## 1. Data loading

```python
# ── Load the cleaned bookings with country names ─────────────────────────────
bookings = pd.read_csv(CLEAN_FILE, keep_default_na=False)
names = pd.read_csv(NAMES_FILE, keep_default_na=False)
bookings = bookings.merge(names[['country', 'country_name']], on='country', how='left')
bookings['arrival_month'] = pd.to_datetime(bookings['arrival_date']).dt.to_period('M').dt.to_timestamp()
print(f'Rows: {len(bookings):,}')
```

## 2. Initial exploration: aggregated tables for the tests

The raw rows are never charted (architecture.md): the tests use small tables aggregated with pandas, at the level the interactions need (month × hotel, and country × hotel × calendar month), so the filters still work on the columns they need.

```python
# ── Month x hotel: guests of stays, cancellation rate and median price ───────
stays = bookings[bookings['is_canceled'] == 0]
by_month = (bookings.groupby(['arrival_month', 'hotel'])
            .agg(bookings_total=('is_canceled', 'size'), cancelled=('is_canceled', 'sum'))
            .join(stays.groupby(['arrival_month', 'hotel']).agg(guests=('total_guests', 'sum')))
            .join(stays[stays['adr'] > 0].groupby(['arrival_month', 'hotel']).agg(median_price=('adr', 'median')))
            .reset_index())
by_month['cancel_rate'] = 100 * by_month['cancelled'] / by_month['bookings_total']
by_month['year'] = by_month['arrival_month'].dt.year
by_month.head()
```

```python
# ── Country x hotel x calendar month: guests of stays (top 10 countries + Other) ─
guests_by_country = stays.groupby('country_name')['total_guests'].sum().sort_values(ascending=False)
top_10 = guests_by_country.drop('Unknown', errors='ignore').head(10).index
stays = stays.assign(country_group=stays['country_name'].where(stays['country_name'].isin(top_10), 'Other'),
                     calendar_month=pd.to_datetime(stays['arrival_date']).dt.month)
by_country_month = (stays.groupby(['country_group', 'calendar_month'])
                    .agg(guests=('total_guests', 'sum'), median_lead_time=('lead_time', 'median')).reset_index())
by_country_month['share'] = 100 * by_country_month['guests'] / by_country_month.groupby('country_group')['guests'].transform('sum')
by_country = stays.groupby(['country_group', 'hotel'], as_index=False)['total_guests'].sum().rename(columns={'total_guests': 'guests'})
by_country['share'] = 100 * by_country['guests'] / by_country.groupby('hotel')['guests'].transform('sum')
print(f'{len(by_month)} rows month x hotel | {len(by_country_month)} rows country x calendar month')
print(f"Months on the time axis: {by_month['arrival_month'].nunique()} | country groups on the country charts: {by_country['country_group'].nunique()}")
by_country.sort_values('guests', ascending=False).head()
```

### How the plan follows Shneiderman's mantra

**Overview first:** each screen opens with data already visible (both hotels, whole period, no country selected) and its KPIs; the Chart / Map switch (I8) and the animation (I10) give other overviews of the same data. **Zoom and filter:** the hotel and year filters (I1, I2), with reset (I4) and the sort toggle (I9). **Details on demand:** tooltips (I6), the linked month (I5), the selected country (I7) and the tables (I11). Interaction has a cognitive cost (course 07), so there is no zoom, pan or drag: the charts have at most 26 months or 12 countries (printed above) and fit the screen. The animation only runs when Natalia presses Play.

## 5. Accessible visualisation: feasibility tests

Each test builds a small version of the chart with the Altair feature the interaction needs, validates it against the Vega-Lite schema (`to_dict`) and compiles it to SVG with vl-convert (the same engine family as the browser). The keyboard part, the overlay focus ring, aria-live and the link parameters are JavaScript around the charts (wcag-accessibility skill) and are built and tested in Part 6.

```python
# ── Helper: validate and compile a chart, return True if both work ───────────
def compiles(chart):
    spec = chart.to_dict()                       # raises an error if the spec is not valid Vega-Lite
    svg = vlc.vegalite_to_svg(spec)              # raises an error if Vega cannot compile it
    return svg.startswith('<svg')

HOTEL_COLOURS = alt.Scale(domain=['City Hotel', 'Resort Hotel'], range=['#0072B2', '#D55E00'])   # Okabe-Ito
SELECTED = '#009E73'
HOTEL_DASHES = alt.Scale(domain=['City Hotel', 'Resort Hotel'], range=[[1, 0], [6, 3]])
tests = {}
print('Helper ready')
```

```python
# ── I1 + I2: hotel and year filters as parameters ────────────────────────────
hotel_param = alt.param(name='hotel_filter', value='Both',
                        bind=alt.binding_radio(options=['Both', 'City Hotel', 'Resort Hotel'], name='Hotel: '))
year_param = alt.param(name='year_filter', value=0,
                       bind=alt.binding_select(options=[0, 2015, 2016, 2017],
                                               labels=['All', '2015 (Jul-Dec)', '2016', '2017 (Jan-Aug)'], name='Year: '))

c1_filtered = alt.Chart(by_month).mark_line().encode(
    x=alt.X('arrival_month:T', title='Month of arrival'),
    y=alt.Y('guests:Q', title='Guests (stays)'),
    color=alt.Color('hotel:N', scale=HOTEL_COLOURS, legend=None),
    strokeDash=alt.StrokeDash('hotel:N', scale=HOTEL_DASHES, legend=None),
).add_params(hotel_param, year_param).transform_filter(
    "(hotel_filter == 'Both' || datum.hotel == hotel_filter) && (year_filter == 0 || datum.year == year_filter)"
).properties(title='I1 + I2: filters', width=400, height=150)

tests['I1 Hotel filter'] = tests['I2 Year filter'] = compiles(c1_filtered)
c1_filtered
```

```python
# ── I5 + I6: linked month highlight across C1-C3, with tooltips ──────────────
import warnings
# The same selection is added to the three charts on purpose (that is what links them),
# so Altair's notice that it merged the identical copies is expected and hidden
warnings.filterwarnings('ignore', message='Automatically deduplicated selection parameter')

month_pick = alt.selection_point(name='picked_month', fields=['arrival_month'], on='pointerover',
                                 nearest=True, empty=False)

def month_chart(field, title):
    base = alt.Chart(by_month).encode(x=alt.X('arrival_month:T', title='Month of arrival'))
    lines = base.mark_line().encode(
        y=alt.Y(f'{field}:Q', title=title),
        color=alt.Color('hotel:N', scale=HOTEL_COLOURS, legend=None),
        strokeDash=alt.StrokeDash('hotel:N', scale=HOTEL_DASHES, legend=None))
    points = base.mark_point(filled=True).encode(
        y=f'{field}:Q', color=alt.Color('hotel:N', scale=HOTEL_COLOURS, legend=None),
        shape=alt.Shape('hotel:N', scale=alt.Scale(domain=['City Hotel', 'Resort Hotel'], range=['circle', 'square']), legend=None),
        tooltip=[alt.Tooltip('arrival_month:T', title='Month', format='%B %Y'), alt.Tooltip('hotel:N', title='Hotel'),
                 alt.Tooltip(f'{field}:Q', title=title, format=',.0f')]).add_params(month_pick)
    rule = base.mark_rule(color='#1A1A1A').transform_filter(month_pick)
    return alt.layer(lines, points, rule).properties(width=400, height=100)

# C2 is a calendar heatmap: the same selection draws an outline around the cells of the month
heatmap_base = alt.Chart(by_month).transform_calculate(
    month_number='month(datum.arrival_month) + 1', row_label="datum.hotel + ' ' + datum.year",
    cancel_class='min(floor(datum.cancel_rate / 10), 6)')
heatmap_cells = heatmap_base.mark_rect().encode(
    x=alt.X('month_number:O', title='Month'), y=alt.Y('row_label:N', title=None),
    color=alt.Color('cancel_class:O', scale=alt.Scale(scheme='purples'), title='Cancelled (class)'),
    tooltip=[alt.Tooltip('arrival_month:T', title='Month', format='%B %Y'), alt.Tooltip('cancel_rate:Q', title='Cancelled (%)', format='.1f')]
).add_params(month_pick)
heatmap_outline = heatmap_base.mark_rect(filled=False, stroke='#1A1A1A', strokeWidth=3).encode(
    x='month_number:O', y='row_label:N').transform_filter(month_pick)
heatmap = alt.layer(heatmap_cells, heatmap_outline).properties(width=400, height=120)

linked = alt.vconcat(month_chart('guests', 'Guests'), heatmap,
                     month_chart('median_price', 'Median price')).properties(title='I5 + I6: linked month and tooltips')
tests['I5 Linked month highlight'] = tests['I6 Tooltips'] = compiles(linked)
linked
```

```python
# ── I7 + I9: select a country (a parameter shared by the charts, also set from a list) and sort toggle ─
countries = sorted(by_country['country_group'].unique())
selected_country = alt.param(name='selected_country', value='',          # '' = no country selected
                             bind=alt.binding_select(options=[''] + countries, labels=['No country'] + countries, name='Country: '))
sort_param = alt.param(name='sort_order', value='guests',
                       bind=alt.binding_radio(options=['guests', 'alphabetical'], name='Sort: '))
clicked = alt.selection_point(name='clicked_country', fields=['country_group'], on='click', empty=False)
is_selected = 'datum.country_group == selected_country'

dumbbell_base = alt.Chart(by_country).transform_joinaggregate(total='sum(guests)', groupby=['country_group']).transform_calculate(
    sort_key="sort_order == 'guests' ? -datum.total : datum.country_group")
country_axis = alt.Y('country_group:N', title=None, sort=alt.EncodingSortField(field='sort_key', op='min'))
dumbbell = alt.layer(
    dumbbell_base.mark_rule(color='#A6A6A6').encode(y=country_axis, x='min(share):Q', x2='max(share):Q'),
    dumbbell_base.mark_point(filled=True).encode(
        y=country_axis, x=alt.X('share:Q', title="% of the hotel's guests"),
        color=alt.Color('hotel:N', scale=HOTEL_COLOURS, legend=None),
        shape=alt.Shape('hotel:N', scale=alt.Scale(domain=['City Hotel', 'Resort Hotel'], range=['circle', 'square']), legend=None),
        size=alt.when(is_selected).then(alt.value(170)).otherwise(alt.value(90)),
        stroke=alt.when(is_selected).then(alt.value('#1A1A1A')).otherwise(alt.value('white')),
        tooltip=[alt.Tooltip('country_group:N', title='Country'), alt.Tooltip('share:Q', title='% of guests', format='.1f')],
    ).add_params(clicked),
).add_params(selected_country, sort_param).properties(width=250, height=200)

# The same parameter drives a second chart: the box plot of nights, the selected box in green
nights = stays[['country_group', 'stays_in_weekend_nights', 'stays_in_week_nights']].assign(
    nights=lambda t: t['stays_in_weekend_nights'] + t['stays_in_week_nights'])
boxes = alt.Chart(nights).mark_boxplot(extent=1.5, outliers=False).encode(
    y=alt.Y('country_group:N', title=None), x=alt.X('nights:Q', title='Nights'),
    color=alt.when(is_selected).then(alt.value(SELECTED)).otherwise(alt.value('#D9D9D9')),
).properties(width=250, height=200)

country_link = alt.hconcat(dumbbell, boxes).properties(title='I7 + I9: country selection and sort')
tests['I7 Select a country'] = tests['I9 Sort toggle'] = compiles(country_link)
country_link
```

```python
# ── I10: animated bubbles, one frame per calendar month, set by Play (a timer in the page) or a slider ─
frame_month = alt.param(name='frame_month', value=1, bind=alt.binding_range(min=1, max=12, step=1, name='Month: '))
bubbles = alt.Chart(by_country_month).transform_filter(
    "datum.calendar_month == frame_month && datum.country_group != 'Other'"
).mark_circle(opacity=0.85, stroke='#595959').encode(
    x=alt.X('median_lead_time:Q', title='Days booked ahead (median)', scale=alt.Scale(zero=True)),
    y=alt.Y('share:Q', title='Arrivals (% of its year)', scale=alt.Scale(zero=True)),
    size=alt.Size('guests:Q', legend=None),
    color=alt.when(is_selected).then(alt.value(SELECTED)).otherwise(alt.value('#C8C8C8')),
    tooltip=[alt.Tooltip('country_group:N', title='Country'), alt.Tooltip('guests:Q', title='Guests', format=',d')],
).add_params(frame_month, selected_country).properties(width=400, height=200, title='I10: one month of the animation')
tests['I10 Play the year'] = compiles(bubbles)
bubbles
```

```python
# ── I12: night mode = the same spec with the light colours swapped for the dark ones ─
import json
NIGHT = {'#0072B2': '#56B4E9', '#D55E00': '#E69F00', '#1A1A1A': '#F2F2F2', '#C8C8C8': '#4D4D4D'}
night_text = json.dumps(bubbles.to_dict())
for light, dark in NIGHT.items():
    night_text = night_text.replace(light, dark)
night_spec = json.loads(night_text)
night_spec['background'] = '#1E1E1E'
tests['I12 Night mode'] = vlc.vegalite_to_svg(night_spec).startswith('<svg')
print('Night version compiles:', tests['I12 Night mode'])
```

```python
# ── Results of the feasibility tests ─────────────────────────────────────────
html_only = {'I3 Filters kept between screens': 'HTML link with ?hotel=&year= read by JavaScript',
             'I4 Reset filters': 'HTML button that sets the parameters back with view.signal()',
             'I8 Chart / Map switch': 'HTML radio buttons that show one of two chart boxes (hidden attribute)',
             'I11 Table view': 'HTML button with aria-pressed that swaps the chart for a table built from the same data'}
feasibility = pd.DataFrame(
    [(name, 'compiled to SVG' if ok else 'FAILED',
      'light colours swapped for the night ones in the spec, as the page does' if name.startswith('I12')
      else 'Vega-Altair parameter / selection') for name, ok in tests.items()]
    + [(name, 'no chart feature needed', how) for name, how in html_only.items()],
    columns=['interaction', 'result', 'how'])
feasibility = feasibility.sort_values('interaction', key=lambda names: names.str.extract(r'I(\d+)')[0].astype(int))
print('All chart tests passed:', all(tests.values()))
feasibility
```

**Finding:** every chart feature needed by the plan compiles in Vega-Altair 6.2.2 (Vega-Lite 6.4.1): filters bound to parameters, a nearest-point selection shared by three charts with a rule or an outline, tooltips, a country parameter shared by two charts and set from a list, a sort key computed from a parameter, one frame of the animation chosen by a slider, and the night colours swapped in the spec. The Play button itself is a timer in the page that changes the month parameter. One limit, said honestly: in the dashboard each chart is embedded separately (each needs its own figure, alternative text and data table), so a selection made in one embed is passed to the others with a few lines of JavaScript (`view.signal()` and `addSignalListener`). This is standard vega-embed use, but it is built and tested only in Part 6.
